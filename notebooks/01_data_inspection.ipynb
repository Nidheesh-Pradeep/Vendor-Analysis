```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_dir = Path("/Users/nidheeshpradeepkumar/Downloads/GitHub/Vendor Analysis/Vendor Analysis/data")
csv_files = sorted(data_dir.rglob("*.csv"))
if not csv_files:
    raise FileNotFoundError(
        "No CSV files found in {data_dir.resolve()}. "
        "Put the extracted Kaggle CSV files in data/raw."
    )
```

```python
for file_path in csv_files:
    df = pd.read_csv(file_path, low_memory=False)

    print(f"\n{'=' * 70}")
    print(f"FILE: {file_path}")
    print(f"ROWS: {len(df):,} | COLUMNS: {len(df.columns)}")

    print("\nColumn names and data types:")
    print(df.dtypes.to_string())

    print("\nFirst five rows:")
    display(df.head())

    print("\nMissing values and distinct values:")
    profile = pd.DataFrame({
        "data_type": df.dtypes.astype(str),
        "missing_count": df.isna().sum(),
        "missing_percent": (df.isna().mean() * 100).round(2),
        "distinct_values": df.nunique(dropna=True),
    })
    display(profile)

    print(f"\nExact duplicate rows: {df.duplicated().sum():,}")

    date_columns = [
        column for column in df.columns
        if any(word in column.lower() for word in ["date", "time"])
    ]

    print("\nDate ranges:")
    for column in date_columns:
        parsed_dates = pd.to_datetime(df[column], errors="coerce")
        valid_dates = parsed_dates.dropna()

        if len(valid_dates):
            invalid_dates = (df[column].notna() & parsed_dates.isna()).sum()
            print(
                f"{column}: {valid_dates.min()} to {valid_dates.max()} "
                f"| unparsed nonblank values: {invalid_dates:,}"
            )
        else:
            print(f"{column}: no dates could be parsed")
```

```python
# Exact duplicate rows across all columns
exact_duplicates = df[df.duplicated(keep=False)]
print(f"Exact duplicate rows: {len(exact_duplicates)}")

# POs appearing on multiple rows—inspect these, but don't remove them automatically
po_column = "PO / SO #"
repeated_pos = df[df[po_column].duplicated(keep=False)].sort_values(po_column)
display(repeated_pos)
```

```python
po_col = "PO / SO #"
qty_col = "Line Item Quantity"
value_col = "Line Item Value"

po_summary = (
    df.groupby(po_col, dropna=False)
      .agg(
          line_count=(po_col, "size"),
          total_line_quantity=(qty_col, "sum"),
          total_line_value=(value_col, "sum"),
      )
      .reset_index()
)

display(po_summary.head())
```

```python
exact_duplicates = df[df.duplicated(keep=False)]
display(exact_duplicates)
```

```python
# Run before adding any new columns
duplicate_mask = df.duplicated(keep=False)
duplicate_rows = df[duplicate_mask].copy()

print(f"Rows involved in exact duplicates: {duplicate_mask.sum()}")
print(f"Extra copies beyond the first: {df.duplicated().sum()}")

display(duplicate_rows)
```

```python
multi_line_pos = po_summary[po_summary["line_count"] > 1]
display(multi_line_pos)
```

```python
multi_line_ids = multi_line_pos["PO / SO #"]

display(
    df[df["PO / SO #"].isin(multi_line_ids)]
      .sort_values("PO / SO #")
)
```

```python
print(f"Total source rows: {len(df):,}")
print(f"POs with multiple lines: {len(multi_line_pos):,}")

multi_line_ids = multi_line_pos["PO / SO #"]
rows_for_multi_line_pos = df[df["PO / SO #"].isin(multi_line_ids)]
print(f"Rows belonging to multi-line POs: {len(rows_for_multi_line_pos):,}")

display(
    multi_line_pos
      .sort_values("line_count", ascending=False)
      .head(10)
)
```
